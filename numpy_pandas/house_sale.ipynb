```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from keras.src.utils.summary_utils import highlight_symbol
from matplotlib import rcParams
from soupsieve.util import lower
from winerror import IORING_E_COMPLETION_QUEUE_TOO_BIG

rcParams["font.sans-serif"]=["SimHei"]
import seaborn as sns
df=pd.read_csv("data/house_sales.csv")

data=df.dropna()
Q1=data["price"].quantile(0.25)
Q3=data["price"].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
high_bound=Q3+1.5*IQR
data[(data["price"]>=lower_bound)&(data["price"]<=high_bound)]
data["date"]=pd.to_datetime(data['date'])

data["age"]=data['date'].dt.year-data['yr_built']
data["is_renovated"]=data["yr_renovated"].apply(lambda x:1 if x>0 else 0)
numeric_columns=data.select_dtypes(include=[np.number]).columns
data[numeric_columns].describe()
correlation=data[numeric_columns].corr()
```

```python
data.groupby("zipcode")[["price","sqft_living","bedrooms"]].mean()
data.pivot_table(values=["price","sqft_living","bedrooms"],index="zipcode",aggfunc="mean")
data['age_group']=pd.cut(data['age'],5)
age_stats=data.groupby("age_group")[["price","sqft_living","bedrooms"]].mean()
age_stats
```
